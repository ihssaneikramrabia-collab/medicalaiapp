```python

# Dataset : Titanic


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

# Chargement du dataset
data = sns.load_dataset("titanic")

print("Dataset chargé avec succès !")
print("Dimensions :", data.shape)
```

```python

# Q1 & Q2 : EXPLORATION INITIALE


print("=== 1. Les 5 premières lignes ===")
display(data.head(5))

print("\n=== 2. Dimensions du dataset ===")
print("Nombre de lignes :", data.shape[0])
print("Nombre de colonnes :", data.shape[1])

print("\n=== 3. Noms des colonnes ===")
print(data.columns.tolist())
```

```python

# Q3 : INFORMATIONS GÉNÉRALES ET TYPES


print("=== Informations générales ===")
data.info()

print("\n=== Types de données ===")
print(data.dtypes)
```

```python

# Q4 : STATISTIQUES DESCRIPTIVES


print("=== Statistiques des variables numériques ===")
display(data.describe())

print("\n=== Statistiques des variables catégorielles ===")
display(data.describe(include=['object', 'category', 'bool']))
```

```python

# Q5 : SÉPARATION DES VARIABLES


# Variables numériques
var_numeriques = data.select_dtypes(
    include=[np.number]
).columns.tolist()

# Variables catégorielles
var_categorielles = data.select_dtypes(
    include=['object', 'category', 'bool']
).columns.tolist()

print("Variables numériques :")
print(var_numeriques)

print("\nVariables catégorielles :")
print(var_categorielles)

# Identification des variables descriptives et informatives
informations_descriptives = [
    'class',
    'embark_town',
    'deck',
    'who'
]

print("\nVariables descriptives :")
print(informations_descriptives)

# Variables ayant un rôle particulier
print("\nVariable cible : survived")
print("Variable redondante avec survived : alive")
print("Variables numériques codées : pclass et survived")

# Remarque :
# Le dataset ne contient pas d'identifiant unique explicite
# permettant d'identifier chaque passager.
```

```python

# Q6 : VALEURS DISTINCTES ET EFFECTIFS


print("=== Nombre de valeurs distinctes par colonne ===")
print(data.nunique())

print("\n=== Modalités et effectifs ===")

colonnes_a_examiner = ['sex', 'embarked', 'class']

for col in colonnes_a_examiner:

    print(f"\nModalités de la variable '{col}' :")

    print(data[col].value_counts(dropna=False))
```

```python

# Q7 : VALEURS MANQUANTES


print("=== Analyse des valeurs manquantes ===")

val_manquantes = pd.DataFrame({

    'Nombre': data.isnull().sum(),

    'Pourcentage (%)':
    (data.isnull().sum() / len(data)) * 100

})

# Afficher uniquement les colonnes contenant des valeurs manquantes
display(
    val_manquantes[val_manquantes['Nombre'] > 0]
)

print("\nNombre total de valeurs manquantes :")
print(data.isnull().sum().sum())
```

```python

# Q8 : RECHERCHE DES DOUBLONS


print("=== Recherche des doublons ===")

nb_doublons = data.duplicated().sum()

print("Nombre de lignes dupliquées :", nb_doublons)

if nb_doublons > 0:

    print("\nAperçu des lignes dupliquées :")

    display(
        data[data.duplicated(keep=False)].head(10)
    )

else:

    print("Aucun doublon trouvé.")

# Décision :
# On conserve les doublons pour le moment.
# En effet, deux passagers peuvent avoir des informations identiques.
# Le dataset ne possède pas d'identifiant unique pour les distinguer.
```

```python

# Q9 : VÉRIFICATION DES INCOHÉRENCES


print("=== Vérification des valeurs incohérentes ===")

# Recherche des âges négatifs
ages_negatifs = data[data['age'] < 0]

# Recherche des tarifs négatifs
tarifs_negatifs = data[data['fare'] < 0]

print("Nombre d'âges négatifs :", len(ages_negatifs))

print("Nombre de tarifs négatifs :", len(tarifs_negatifs))

# Affichage des lignes concernées
if len(ages_negatifs) > 0:
    display(ages_negatifs)

if len(tarifs_negatifs) > 0:
    display(tarifs_negatifs)

print("\nConclusion :")
print("Les valeurs négatives doivent être considérées comme incohérentes.")
print("Les valeurs inhabituelles doivent être vérifiées avant leur suppression.")
```

```python

# Q10 : HARMONISATION DES CHAÎNES DE CARACTÈRES


print("=== Nettoyage des variables textuelles ===")

# Sélection des variables textuelles
colonnes_textuelles = data.select_dtypes(
    include=['object', 'category', 'string']
).columns

for col in colonnes_textuelles:

    # Conversion en type string
    data[col] = data[col].astype('string')

    # Suppression des espaces superflus
    data[col] = data[col].str.strip()

    # Conversion en minuscules
    data[col] = data[col].str.lower()

    # Transformation des chaînes vides en valeurs manquantes
    data[col] = data[col].replace('', pd.NA)

print("Nettoyage effectué avec succès !")

print("\n=== Vérification des chaînes vides ===")

for col in colonnes_textuelles:

    nb_vides = (data[col] == '').sum()

    print(f"{col} : {nb_vides} chaîne(s) vide(s)")

print("\n=== Vérification des valeurs manquantes ===")
print(data.isnull().sum())
```

```python

# Q11 : JUSTIFICATION DU NETTOYAGE


print("=== Justification des opérations de nettoyage ===")

print("""
1. Les espaces superflus sont supprimés pour uniformiser les données.

2. Les chaînes sont converties en minuscules pour éviter
   les différences de casse.

3. Les chaînes vides sont transformées en valeurs manquantes.

4. Les doublons sont examinés avant toute suppression.

5. Les valeurs négatives sont considérées comme incohérentes
   et doivent être vérifiées.

6. Les valeurs inhabituelles ne sont pas supprimées automatiquement.
""")
```

```python

# Q12 : STRATÉGIE POUR LES VARIABLES NUMÉRIQUES


print("=== Traitement des variables numériques ===")

colonnes_numeriques = data.select_dtypes(
    include=[np.number]
).columns

for col in colonnes_numeriques:

    nb_manquantes = data[col].isnull().sum()

    if nb_manquantes > 0:

        print(f"\nVariable : {col}")

        print("Nombre de valeurs manquantes :", nb_manquantes)

        print("Stratégie retenue : remplacement par la médiane.")

print("""
Justification :
La médiane est choisie car elle est moins sensible
aux valeurs extrêmes (outliers) que la moyenne.
""")
```

```python

# Q13 : STRATÉGIE POUR LES VARIABLES CATÉGORIELLES


print("=== Traitement des variables catégorielles ===")

colonnes_categorielles = data.select_dtypes(
    include=['object', 'category', 'bool', 'string']
).columns

for col in colonnes_categorielles:

    nb_manquantes = data[col].isnull().sum()

    if nb_manquantes > 0:

        pourcentage = (nb_manquantes / len(data)) * 100

        print(f"\nVariable : {col}")

        print("Valeurs manquantes :", nb_manquantes)

        print("Pourcentage :", round(pourcentage, 2), "%")

        if pourcentage > 70:

            print("Stratégie proposée : catégorie Inconnu")

        else:

            print("Stratégie proposée : remplacement par le mode.")

print("""
Comparaison des solutions :

1. Mode :
   Remplacer les valeurs manquantes par la modalité la plus fréquente.

2. Catégorie Inconnu :
   Ajouter une catégorie pour représenter les valeurs absentes.

3. Conservation des valeurs manquantes :
   Garder les valeurs NaN pour certaines analyses.
""")
```

```python

# Q14 : COMPARAISON DES MÉTHODES DE TRAITEMENT


methodes = pd.DataFrame({

    'Méthode': [
        'Moyenne',
        'Médiane',
        'Mode',
        'Valeur constante',
        'Suppression'
    ],

    'Utilisation': [

        'Pour les variables numériques sans valeurs extrêmes importantes.',

        'Pour les variables numériques contenant des valeurs extrêmes.',

        'Pour les variables catégorielles ou discrètes.',

        'Lorsqu’une valeur par défaut ou une catégorie Inconnu est pertinente.',

        'Lorsque le nombre de valeurs manquantes est faible et leur suppression justifiée.'
    ]

})

display(methodes)
```

```python

# Q15 : APPLICATION DU NETTOYAGE FINAL


# Création d'une copie du DataFrame
df_clean = data.copy()


# 1. Traitement des variables numériques par la médiane


print("=== Traitement des variables numériques ===")

colonnes_numeriques = df_clean.select_dtypes(
    include=[np.number]
).columns

for col in colonnes_numeriques:

    if df_clean[col].isnull().sum() > 0:

        mediane = df_clean[col].median()

        df_clean[col] = df_clean[col].fillna(mediane)

        print(f"{col} : valeurs remplacées par la médiane = {mediane}")



# 2. Traitement des variables catégorielles


print("\n=== Traitement des variables catégorielles ===")

colonnes_categorielles = df_clean.select_dtypes(
    include=['object', 'category', 'bool', 'string']
).columns

for col in colonnes_categorielles:

    nb_manquantes = df_clean[col].isnull().sum()

    if nb_manquantes > 0:

        pourcentage = (nb_manquantes / len(df_clean)) * 100

        if pourcentage > 70:

            # Catégorie Inconnu pour les variables
            # ayant un taux élevé de valeurs manquantes

            df_clean[col] = df_clean[col].fillna('inconnu')

            print(f"{col} : remplacement par 'inconnu'")

        else:

            # Remplacement par le mode

            mode = df_clean[col].mode()

            if len(mode) > 0:

                valeur_mode = mode[0]

                df_clean[col] = df_clean[col].fillna(valeur_mode)

                print(f"{col} : remplacement par le mode = {valeur_mode}")



# 3. Vérification finale des valeurs manquantes


print("\n=== Valeurs manquantes après traitement ===")

print(df_clean.isnull().sum())

print("\nNombre total de valeurs manquantes restantes :")

print(df_clean.isnull().sum().sum())


# 4. Dimensions finales


print("\n=== Dimensions du dataset nettoyé ===")

print("Nombre de lignes :", df_clean.shape[0])

print("Nombre de colonnes :", df_clean.shape[1])



# 5. Sauvegarde du dataset nettoyé

df_clean.to_csv(
    "titanic_nettoye_tp1.csv",
    index=False
)

print("\nFichier 'titanic_nettoye_tp1.csv' enregistré avec succès !")

print("Dataset prêt pour le TP2.")
```
